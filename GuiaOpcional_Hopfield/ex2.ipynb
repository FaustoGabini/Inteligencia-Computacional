```python
def funcion_signo(x): 
  return np.where(x > 0, 1, -1)
```

```python
def entrenar_hebbiano(patrones): 
  N = patrones.shape[1]
  W = np.zeros((N, N))

  for patron in patrones:
    patron = np.reshape(patron, (1, N))
    W += np.dot(patron.T, patron)

np.fill_diagonal(W, 0)

W
```

```python
def recuperar(W, x, max_iter=10): 
  for _ in range(max_iter): 
    x_new = funcion_signo(np.dot(x, W))
    
    if np.array_equal(x, x_new):
      break
    x = x_new
  return x
```

```python
# Probar con cada patrón original (deberían recuperarse a sí mismos)

W = entrenar_hebbiano(patrones)
# Probar con cada patrón original (deberían recuperarse a sí mismos)
for i, x in enumerate(patrones):
    recuperado = recuperar(W, x.copy())
    print(f"Patrón {i} recuperado")
```
