```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from tabulate import tabulate

from KMeans import KMeans

from cohesion_media import cohesion_media
from sklearn.metrics.cluster import contingency_matrix
from sklearn.metrics.cluster import rand_score
from sklearn.metrics.cluster import fowlkes_mallows_score
from sklearn.metrics.cluster import davies_bouldin_score
```

```python

data = pd.read_csv('./data/iris81_trn.csv')
X = data.values.astype(float)
X_entradas = X[:, :-3]  # Todas las columnas excepto las 3 últimas
y = np.argmax(X[:, -3:], axis=1)  # La clase es el índice del valor máximo en las últimas 3 columnas
```

```python
matrix = []
db = []
ri = []
fm = []
for k in range(2, 11): 
    k_means = KMeans(k, 30)
    k_means.fit(X_entradas)
    matrix.append(contingency_matrix(y, k_means.labels))
    db.append(davies_bouldin_score(X_entradas, k_means.labels))
    ri.append(rand_score(k_means.labels, y))
    fm.append(fowlkes_mallows_score(k_means.labels, y))
```

```python
# Tabla con matriz de contingencia, Davies-Bouldin, Rand Index y Fowlkes & Mallows
simple_table_data = [[k, tabulate(mat, tablefmt='grid'), db_score, ri_score, fm_score] for k, mat, db_score, ri_score, fm_score in zip(range(2,11), matrix, db, ri, fm)]
simple_headers = ['k','Matriz de Contingencia','Davies-Bouldin','Rand Index','Fowlkes & Mallows']
simple_table = tabulate(simple_table_data, simple_headers, tablefmt='simple_grid',stralign='center',numalign='center')
print(simple_table)
```

## 📊 Davies-Bouldin Index

El **Davies-Bouldin Index (DBI)** es una métrica **interna** de evaluación de clustering.

- **No requiere etiquetas reales.**
- Mide la **compacidad** de los clusters y la **separación entre ellos**.
- Un valor **más bajo** indica mejor clustering.

- Penaliza si los clusters están muy dispersos (mal definidos).
- Penaliza si están muy cerca entre sí (mal separados).

- Cuanto **menor**, mejor.

## 📊 Rand Index

El **Rand Index (RI)** es una métrica **externa** para comparar un clustering con las clases reales.

- Compara todos los **pares de puntos** posibles.
- Cuenta cuántos están correctamente agrupados (juntos o separados).
- **Sí requiere etiquetas verdaderas** (`y_true`).

- Mide si el clustering respeta las relaciones que existen en la verdad.
- **1.0** = agrupamiento perfecto
- **0.0** = completamente aleatorio


## 📊 Fowlkes & Mallows Score

El **Fowlkes-Mallows Score (FMI)** es una métrica **externa** basada en pares de puntos.

- Mide la **precisión** y **recall** entre pares de puntos en los clusters.
- Es la **media geométrica** entre:
  - Precisión: qué tan correctos son los pares agrupados.
  - Recall: cuántos pares correctos fueron encontrados.

- **1.0** = agrupamiento perfecto
- **0.0** = agrupamiento aleatorio

- Penaliza tanto los falsos positivos como los falsos negativos.
